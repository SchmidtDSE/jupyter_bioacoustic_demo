# ANNOTATE

Annotation workflow: assign species and mark time/frequency regions. Arguments can be passed directly to the `BioacousticAnnotator`, however it is often much more convient to use configuration files.  Here, in addition to the annotation files, we have passed additional kwargs (annotator_name, annotation_date) which will be passed directly to the output data.

```python
ba_annotate = BioacousticAnnotator(
    project='annotator_config/projects/simple_annotator.yaml',
    annotator_name='brookie',
    annotate_date=20260515
)
```

```python
ba_annotate.open()
```

```python
print_md('**Source Data**')
ba_annotate.source.sample(3)
```

```python
print_md('**Annotated Data**')
ba_annotate.output().sample(3)
```
